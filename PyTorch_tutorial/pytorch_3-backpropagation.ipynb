## Backpropagation

Consider a computational graph
$$(x,w) \xrightarrow{*} \hat{y} \xrightarrow[y\uparrow ]{-} s \xrightarrow{\hat{L}} \text{loss}$$
where we want to use linear regression. We mopartial our output as a linear combination of some weights and an input ($\hat{y}$ is the predicted $y$):
$$\hat{y} = wx$$
and then we formulate some loss function, in this case we use the squared error (usually you take MSE)
$$\text{loss} = (\hat{y} - y)^2 = (wx - y)^2 = s^2$$
Now we want to minimize our loss, so we want to know 
$$\frac{\partial \text{loss}}{\partial w}.$$
We first do a forward pass to calculate $\hat{y}, s$ and the loss. Then we calculate the local gradients at each node (operation) wrt the weights, so we calculate, in order: 
$$\frac{\partial \text{loss}}{\partial s}, \frac{\partial s}{\partial \hat{y}}, \frac{\partial \hat{y}}{\partial w}.$$
Then we do our backwards pass by making use of the chain rule to calculate 
$$\frac{\partial \text{loss}}{\partial w}.$$

\underline{Example with numbers}: Let us use as training samples $(x,y) = (1, 2)$ and we initialize our weights as $w = 1$. We do the forward pass: $\hat{y} = wx = 1 \cdot 1 = 1$, $s = \hat{y} - y = 1 - 2 = -1$ and then loss $= s^2 = 1$.\par 
Now we calculate the local gradient: 
$$\frac{\partial \text{loss}}{\partial s} = \frac{\partial s^2}{\partial s} = 2s = -2, \frac{\partial s}{\partial \hat{y}} = \frac{\partial (\hat{y}-y)}{\partial \hat{y}} = 1 \text{ and } \frac{\partial \hat{y}}{\partial w} = \frac{\partial wx}{\partial w} = x = 1.$$
Then we do the backward pass: 
$$\frac{\partial \text{loss}}{\partial w} = \frac{\partial \text{loss}}{\partial s}\frac{\partial s}{\partial \hat{y}}\frac{\partial \hat{y}}{\partial w} = -2 \cdot 1 \cdot 1 = -2. $$

Let us now verify this in Python: 

```python
x = torch.tensor(1.0)
y = torch.tensor(2.0)

w = torch.tensor(1.0, requires_grad=True)  # we are interested in the gradient

# forward pass and compute loss
y_hat = w * x 
loss = (y_hat - y)**2
print(loss)

# backward pass (PyTorch will compute local gradients and backward pass automatically)
loss.backward()
print(w.grad.item())

# NEXT: update weights, next forward and backwards pass, etc.
```
