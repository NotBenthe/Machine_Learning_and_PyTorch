## Dataset and DataLoader

Optimizing model based on whole dataset might be very time consuming. A better way for large datasets is to divide the dataset into so-called *batches*, and then you loop over epochs and then over the number of total batches. Then you get the `x_batch` and `y_batch` samples and do the optimization only over those batches. PyTorch can do that using `DataLoader`. 

Definitions: 
    - epoch: 1 forward and backward pass of ALL training samples
    - `batch_size`: number of training samples in one forward \& backward pass
    - number of iterations: number of passes, each pass using `batch_size` number of samples (ex: 100 samples with `batch_size` = 20 $\rightarrow$ $100/20=5$ iterations per epoch)


We have the wine dataset from the tutorial (https://www.youtube.com/watch?v=c36lUUr864M, to find at https://github.com/patrickloeber/pytorchTutorial/blob/master/data/wine/wine.csv). The first row is a header. We want to predict the wine categories (1, 2 or 3), and the class label is in the very first column. All the other columns are the features. 

```python
# impement our own custom dataset 
class WineDataset(Dataset):

    def __init__(self):
        # data loading 
        xy = np.loadtxt('wine.csv', delimiter=',', dtype=np.float32, skiprows=1)
        self.x = torch.from_numpy(xy[:, 1:])
        self.y = torch.from_numpy(xy[:, [0]])  # n_samples, 1
        self.n_samples = xy.shape[0]

    def __getitem__(self, index):
        return self.x[index], self.y[index]

    def __len__(self):
        return self.n_samples

dataset = WineDataset()
first_data = dataset[0]
features, labels = first_data
print(features, labels)
```

```python
batch_size = 4

dataset = WineDataset()
dataloader = DataLoader(dataset=dataset, batch_size=batch_size, shuffle=True, num_workers=2)  # shuffle = True useful for training, 
                                    #increasing num_workers might make it faster since it's using multiple subprocesses now

datatiter = iter(dataloader)
data = next(datatiter)
features, labels = data 
print(features, labels)
```

```python
# dummy training loop 
num_epochs = 2
total_samples = len(dataset)
n_iterations = math.ceil(total_samples / batch_size)
print(total_samples, n_iterations)

for epoch in range(num_epochs):
    for i, (inputs, labels) in enumerate(dataloader):
        # forward, backward, update
         
        if (i+1) % 5 == 0: 
            print(f'epoch {epoch+1}, step {i+1}/{n_iterations}, inputs {inputs.shape}')
```

## Transforms for Datasets

We add a transform in the dataset

```python
# impement our own custom dataset 
class WineDataset(Dataset):

    def __init__(self, transform=None ):
        # data loading 
        xy = np.loadtxt('wine.csv', delimiter=',', dtype=np.float32, skiprows=1)
        self.x = xy[:, 1:]
        self.y = xy[:, [0]] # n_samples, 1
        self.n_samples = xy.shape[0]

        self.transform = transform

    def __getitem__(self, index):
        sample = self.x[index], self.y[index]
        if self.transform: 
            sample = self.transform(sample)
        return sample 

    def __len__(self):
        return self.n_samples
```

We will create our own transform classes, i.e. to transform it to a tensor

```python
class ToTensor:
    def __call__(self, sample): 
        inputs, targets = sample 
        return torch.from_numpy(inputs), torch.from_numpy(targets)

dataset = WineDataset(transform=None)
first_data = dataset[0]
features, labels = first_data
print(features)
print(type(features), type(labels))

dataset = WineDataset(transform=ToTensor())
first_data = dataset[0]
features, labels = first_data
print('\n', features)
print(type(features), type(labels))
```

we can also compose two transforms: 

```python
class MulTransform: 
    def __init__(self, factor):
        self.factor = factor 

    def __call__(self, sample): 
        inputs, target = sample 
        inputs *= self.factor 
        return inputs, target 

factor = 2
composed = torchvision.transforms.Compose([ToTensor(), MulTransform(factor)])
dataset = WineDataset(transform=composed)

first_data = dataset[0]
features, labels = first_data
print(features)
print(type(features), type(labels))
```
