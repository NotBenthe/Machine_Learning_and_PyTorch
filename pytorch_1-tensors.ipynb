## Tensors

```python
# creating an empty tensor 
size = 3
x = torch.empty(size)
print(x)

size = (2, 3)
x = torch.empty(size)
print(x)

# creating tensor with random values 
x = torch.rand(size)
print(x)

# creating tensor with zeros/ones
x = torch.zeros(size)
print(x)
x = torch.ones(size)
print(x)
```

The default datatype for a PyTorch tensor is `float32`, but we can also assign it a different datatype: 

```python
x = torch.ones(size)
print(x, x.dtype)

x = torch.ones(size, dtype=int)  # or float16, double, float32
print(x, x.dtype)

# seeing size
print(x.size())
```

making tensor from list

```python
x = torch.tensor([2.5, 0.1])
print(x)
```

basic operations

```python
x, y = torch.rand(2, 2), torch.rand(2, 2)

print(x, '\n', y)

print('\naddition')
z = x + y 
print(z)
z = torch.add(x, y)
print(z)

# add elements of x to y
print('\nadding elements')
y.add_(x)
print(y)

# subtracting 
z = x - y 
z = torch.sub(x, y)
y.sub_(x)

# element wise multiplication 
z = x * y
z = torch.mul(x, y)
y.mul_(x)

# element wise division
z = x / y
z = torch.div(x,y)
y.div_(x)
```

indexing and reshaping

```python
x = torch.rand(5, 3) 
print(x)

# rows, columns
print(x[:, 0])

print(x[1, :])

# for tensor with one element to get actual value
print(x[1, 1])
print(x[1, 1].item())
```

```python
x = torch.rand(4, 4)

# reshaping
y = x.view(16)  # number of elements must be the same
print(y)

y = x.view((4, 4))
print(y)

# automatically determining right size
y = x.view(-1, 8)
print(y.shape)
print(y.size())
```

converting to/from numpy

```python
# tensor to numpy
a = torch.ones(5)
print(a.type())
b = a.numpy()
print(type(b))

# numpy to tensor
a = np.ones(5)
print(type(a))
b = torch.from_numpy(a)
print(b.type())

# if tensor is on CPU and not on the GPU,, both objects are on the same memory slice 
# so changing one also changes the other: 
b[1] = 2
print(a)
print(b)

# solution: 
print('\nsolution')
a = torch.ones(5)
b = a.clone().numpy()

b[1] = 2
print(a)
print(b)

a = np.ones(5)
b = torch.from_numpy(a).clone()

b[1] = 2
print(a)
print(b)
```

`requires_grad`: by default `False`, if `True` it tells PyTorch that it will need to calculate the gradient of the tensor later in the optimization steps. 

```python
x = torch.ones(5, requires_grad=True)
```
