## Gradient Descent using Autograd

1. We start by applying linear regression algorithm from scratch (every step manually): implement equations to calculate model prediction and loss function, then numerically calculate the gradients and then update the parameters by gradient descent. 

2. We can then use autograd for gradients computation.

3. We replace calculation of loss computation and parameter updates by PyTorch Loss and Optimizer respectively. 

4. Implement PyTorch Model instead of manual model. 

### Step 1

```python
# linear regression, we don't care about bias here
# f = w * x 
# example: w = 2

# training samples
X = np.array([1, 2, 3, 4], dtype=np.float32)
Y = np.array([2, 4, 6, 8], dtype=np.float32) # since Y = 2 * X

# weight initialization 
w = 0.0 

# model prediction
def forward(x): 
    return w * x 

# loss = MSE (in case of linear regression)
def loss(y, y_pred): 
    return ((y-y_pred)**2).mean()

# gradient
# MSE = 1/N * (w*x - y)**2 
# dJ/dw = 1/N * 2x * (w*x - y)
def gradient(x, y, y_pred):
    return np.dot(2*x, y_pred - y).mean()

print(f'prediction before training: f(5) = {forward(5):.3f}')

# training 
learning_rate = 0.01 
n_iter = 10 

# training loop 
for epoch in range(n_iter): 
    # forward pass
    y_pred = forward(X)

    # loss
    l = loss(Y, y_pred)

    # gradient
    dw = gradient(X, Y, y_pred)

    # update our weights (update formula in gradient descent)
    w -= learning_rate * dw 

    if epoch % 1 == 0:
        print(f'epoch {epoch+1}: w = {w:.3f}, loss = {l:.8f}')

print(f'prediction after training: f(5) = {forward(5):.3f}')
```

Now we do the gradients with PyTorch

```python
# training samples
X = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
Y = torch.tensor([2, 4, 6, 8], dtype=torch.float32) # since Y = 2 * X

# weight initialization 
w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

# model prediction
def forward(x): 
    return w * x 

# loss = MSE (in case of linear regression)
def loss(y, y_pred): 
    return ((y-y_pred)**2).mean()

print(f'prediction before training: f(5) = {forward(5):.3f}')

# training 
learning_rate = 0.01 
n_iter = 20 

# training loop 
for epoch in range(n_iter): 
    # forward pass
    y_pred = forward(X)

    # loss
    l = loss(Y, y_pred)

    # gradient = backward pass 
    l.backward()
    
    # update our weights (update formula in gradient descent)
    # remember: should not be part of gradient tracking path
    with torch.no_grad():
        w -= learning_rate * w.grad.item() 

    # zero the gradients
    w.grad.zero_()

    if epoch % 2 == 1:
        print(f'epoch {epoch+1}: w = {w:.3f}, loss = {l:.8f}')

print(f'prediction after training: f(5) = {forward(5):.3f}')
```
