## Training Pipeline (Model/Loss/Optimizer)

### General training pipeline in PyTorch

1. Design model (number of inputs/outputs, forward pass)
2. Construct loss and optimizer
3. Training loop
    - forward pass: compute prediction
    - backward pass: gradients
    - update weights

```python
# training samples
X = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
Y = torch.tensor([2, 4, 6, 8], dtype=torch.float32) # since Y = 2 * X

# weight initialization 
w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

# model prediction
def forward(x): 
    return w * x 

print(f'prediction before training: f(5) = {forward(5):.3f}')

# training 
learning_rate = 0.01 
n_iter = 100

# loss function and optimizer
loss = nn.MSELoss()
optimizer = torch.optim.SGD([w], lr=learning_rate)

# training loop 
for epoch in range(n_iter): 
    # forward pass
    y_pred = forward(X)

    # loss (MSE loss)
    l = loss(Y, y_pred)

    # gradient = backward pass 
    l.backward()

    # update weights / do optimization step 
    optimizer.step()
    
    # zero the gradients
    optimizer.zero_grad()

    if epoch % 10 == 9:
        print(f'epoch {epoch+1}: w = {w:.3f}, loss = {l:.8f}')

print(f'prediction after training: f(5) = {forward(5):.3f}')
```

Replace manual forward method with a PyTorch model. `nn.Linear()` needs an input size and output size, so we need to do some modifications. Hence, `X` and `Y` can have a different shape. `X` must be a 2D array now, where the number or rows is the number of samples and for each row we have the features. 

```python
# training samples
X = torch.tensor([[1], [2], [3], [4]], dtype=torch.float32)
Y = torch.tensor([[2], [4], [6], [8]], dtype=torch.float32) # since Y = 2 * X

n_samples, n_features = X.shape
print(f'we have {n_samples} samples and {n_features} features for each sample')

# model prediction, for linear progression
# input size = n_features
# output size = n_features
model = nn.Linear(n_features, n_features)

X_test = torch.tensor([5], dtype=torch.float32)
print(f'prediction before training: f(5) = {model(X_test).item():.3f}')

# training 
learning_rate = 0.01 
n_iter = 100

# loss function and optimizer
loss = nn.MSELoss()

# call parameters function for the weights
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# training loop 
for epoch in range(n_iter): 
    # forward pass
    y_pred = model(X)

    # loss (MSE loss)
    l = loss(Y, y_pred)

    # gradient = backward pass 
    l.backward()

    # update weights / do optimization step 
    optimizer.step()
    
    # zero the gradients
    optimizer.zero_grad()

    if epoch % 10 == 9:
        [w, b] = model.parameters()
        print(f'epoch {epoch+1}: w = {w[0][0].item():.3f}, loss = {l:.8f}')

print(f'prediction after training: f(5) = {forward(X_test).item():.3f}')
```

Custom linear regression model

```python
# training samples
X = torch.tensor([[1], [2], [3], [4]], dtype=torch.float32)
Y = torch.tensor([[2], [4], [6], [8]], dtype=torch.float32) # since Y = 2 * X

n_samples, n_features = X.shape
print(f'we have {n_samples} samples and {n_features} features for each sample')

# model prediction, for linear progression
class LinearRegression(nn.Module):

    def __init__(self, input_dim, output_dim):
        super(LinearRegression, self).__init__()

        # define layers 
        self.lin = nn.Linear(input_dim, output_dim)

    def forward(self, x): 
        return self.lin(x)

model = LinearRegression(n_features, n_features)


X_test = torch.tensor([5], dtype=torch.float32)
print(f'prediction before training: f(5) = {model(X_test).item():.3f}')

# training 
learning_rate = 0.01 
n_iter = 100

# loss function and optimizer
loss = nn.MSELoss()

# call parameters function for the weights
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# training loop 
for epoch in range(n_iter): 
    # forward pass
    y_pred = model(X)

    # loss (MSE loss)
    l = loss(Y, y_pred)

    # gradient = backward pass 
    l.backward()

    # update weights / do optimization step 
    optimizer.step()
    
    # zero the gradients
    optimizer.zero_grad()

    if epoch % 10 == 9:
        [w, b] = model.parameters()
        print(f'epoch {epoch+1}: w = {w[0][0].item():.3f}, loss = {l:.8f}')

print(f'prediction after training: f(5) = {forward(X_test).item():.3f}')
```
